```python
import numpy as np
import pandas as pd
data = {'Name': ['Shrey', 'Gini', 'Chitra', 'David', 'Meher', 'Faizal','Geet'],
        'Age':[15, 16, 17, 15, 17, 16, 15],
        'Interest':['Sketching','Outdoor Games','Cooking','Surfing','Reading','Sketching','Surfing']}
df = pd.DataFrame(data)
print(df)
print(df.Age.unique())
print(f"Mean age: {df.Age.mean():.2f}")
print("The first two row of the data frame: \n", df.head(2))
print("The last two row of the data frame: \n", df.tail(2))
print(df.axes)
df = df.T
print('----')
print(df.axes)
print(df)
print("Is df empty: ",df.empty)
print("Dimension of df: ", df.ndim)
print("Shape of df: ", df.shape)
print("Indexes of df: ", df.index)
print("Values of df: ", df.values)
print("The number of elements: ", df.size)
```

```python
#The output of pivot will be errors if there is any duplicate items
import pandas as pd
df = pd.DataFrame({"Name": ["Gitisha","Sonakshi","Meena"], "Degree":["Masters","Graduate","Graduate"],"Age":[24, 26, 25]})
print("Original Data Frame")
print(df)
print("Data frame after using age and name as values: ")
df1 = df.pivot(index="Name", columns="Degree", values=["Age"])
df2 = df.pivot(index="Name", columns="Degree", values=["Name","Age"])
print(df1)
print(df2)
```

```python
# if there is no corresponding data in pivot, the output will be NaN.
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
'foo': ['one', 'two', 'three', 'four', 'five', 'six'],
'bar': ['A', 'B', 'C', 'D', 'E', 'F'],
'baz': [1, 2, 3, 4, 5, 6],
'zoo': ['x', 'y', 'z', 'q', 'w', 't']
})
print(df)
# Pivot the DataFrame
pivoted_df = df.pivot(index='foo', columns='bar', values='baz')
print(pivoted_df)
```

```python
#sorting
from pandas import DataFrame
Cars = {"Brand": ["Ecosport", "Brezza","Audi Q5", "Audi A5"],
        "Price": [963000, 734000, 5026000, 734000],
        "Year": [2012, 2016, 2008, 2017]}
df = DataFrame(Cars, columns = ["Brand", "Price", "Year"])
print(df)
df.sort_values(by=["Year","Price"], inplace=True)
print("\nSorting....")
print(df)
```

```python
#missing data
import pandas as pd
import numpy as np
df = pd.DataFrame({"one":pd.Series([1,2,3],index=['a','b','c']), "two":pd.Series([4,5,6,7],index=['a','b','c','d']), "three":pd.Series([8, 9, 10], index = ['b','c','d'])})
print(df)
df2= pd.DataFrame({"one":pd.Series([1,2,3],index=['a','c','d']), "two":pd.Series([4,6,7],index=['a','b','c']), "three":pd.Series([8, 9, 10],index=['a','b','d'])})
print(df2)
# adding int + 'NaN' = 'NaN'
print(df.add(df2))
print(df.add(df2, fill_value=0)) #replace 'NaN' with 0 before add to df

```

```python
#adding new indexes and filling missing values
import pandas as pd
import numpy as np
cols = ['c1','c2','c3','c4','c5']
index = ['r1','r2','r3','r4','r5']
df = pd.DataFrame(np.random.rand(5,5), columns=cols, index=index)
print(df)
new_index = ['r1','b','r3','d','r4']
new_df = df.reindex(new_index, fill_value=0.00)
print(new_df)
new_cols = ['a','c2','c3','b','c4']
print(df.reindex(new_cols,axis=1, fill_value = 0.00))
```

```python
import pandas as pd
import numpy as np
import math

def add(num1, num2):
  return(num1+num2)

Test_marks = {"marks_maths":pd.Series([88,90, 92, 67, 73, 59]), "mark_science":pd.Series([65, 67, 78, 77, 89, 90])}
df = pd.DataFrame(Test_marks)
print("original marks")
print(df)
#add 5 to all marks
print("add 5 to all marks")
print(df.pipe(add, 5))
```

```python
# modified all values in the data frame

def reduce(num):
  return (num-2)
print("original marks")
print(df)
print("subtract 2 from all marks")
print(df.apply(reduce))  # can also use df.pipe(reduce)
```

```python
import pandas as pd
data = {"Team": ["Bhabha","Bhabha","Bhabha","Bhabha","Bhabha","Bhabha","Bhabha","Aryabhatta","Aryabhatta","Aryabhatta","Aryabhatta","Aryabhatta","Aryabhatta","Aryabhatta"],
        "Status":["Completed","Completed","Not Completed","Completed","Not Completed","Completed", "Not Completed","Completed","Completed","Not Completed","Not Completed",
                  "Completed","Not Completed","Completed"],
        "Score":[1,9,5,8,2, 7, 5, 6, 4, 5, 7, 8, 3, 9]}
df = pd.DataFrame(data)
print(df)
group2 = df.groupby(["Team","Status"]).agg({"Score":["mean","max","min"]})
print("group 2")
print(group2)
```

```python
#Transform function
import pandas as pd
df = pd.DataFrame ({"A":[12, 4, 5, None, 1],
                    "B":[7, 2, 54, 3, None],
                    "C":[20, 16, 11, 3, 8],
                    "D":[14, 3, None, 2, 6]})
df.index = ["row1", "row2", "row3", "row4", "row5"]
print("original data")
print(df)
result = df.transform(func = ["sqrt","exp"])
print("after transform")
print(result)
```

```python
#reindex
import pandas as pd
import numpy as np
cols = ["c1","c2","c3","c4","c5"]
index = ["r1","r2","r3","r4","r5"]
df = pd.DataFrame(np.random.rand(5,5), columns=cols, index=index)
print("original data")
print(df)

#reindex
print("\nafter reindex")
print(df.reindex(["r3","r1","r2","r4","r5"]))

#newindex
new_index = ["r1","b","r3","d","r4"]
new_df = df.reindex(new_index, fill_value=0.00)
print("\nafter new index")
print(new_df)
```

```python
#data wrangling
import pandas as pd
import numpy as np
data1 = {"RollNo":['01','02','03','04'],
         "Name":["Jai","Princi","Gaurav","Anuj"],
         "Marks":[27, 24, 22, 32]}
data2 = {"RollNo":['01','02','03','04'],
         "Course":["BCA","BTech","MCA","MTech"],
         "Fees":[128000, 140000, 85000, 160000]}
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
print("data1")
print(df1)
print("data2")
print(df2)
res = pd.merge(df1, df2, on="RollNo")
print("after merge")
print(res)
```

```python
#time series data
import pandas as pd
import numpy as np

print(pd.Timestamp('2012-05-01'))
print("date after 1587687255 seconds")
print(pd.Timestamp(1587687255, unit='s'))

#use date as index
dates = pd.DatetimeIndex(["2020-07-04","2021-08-04","2020-09-04","2021-10-4"]) #yyyy-mm-dd
sales = pd.Series([2000, 2500, 4000, 8000], index = dates)
print("\nsales")
print(sales)

print("\nSale in 2021")
print(sales["2021"])
```
